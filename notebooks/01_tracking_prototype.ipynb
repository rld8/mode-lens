# Tracking prototype
This notebook inspects the public tracker adapter on the generated demo. Production algorithms remain in `src/`; the notebook contains no divergent implementation.

```python
from pathlib import Path
from modelens.adapters.opencv_tracking import BrightContourTracker
from modelens.adapters.opencv_video import OpenCVVideoSource
from modelens.config import load_config
from modelens.ports.tracking import TrackingRequest

config = load_config(Path('../configs/demo_cantilever.yaml'))
video = Path('../data/samples/cantilever_baseline.mp4')
source = OpenCVVideoSource()
request = TrackingRequest(config.experiment.axis_start_px, config.experiment.axis_end_px, config.experiment.roi, config.tracking)
tracking = BrightContourTracker().track(source.frames(video), request)
tracking.positions_px.shape, tracking.valid.mean()
```
